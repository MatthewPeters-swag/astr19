```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 1, 100)

def my_exp(x):
    return np.exp(x)
```

```python
y = my_exp(x)

# Plot
plt.figure()
plt.plot(x, y)

# Labels
plt.xlabel("Time [milliseconds]")
plt.ylabel("Awesomeness")

# Save as PDF
plt.savefig("exp_plot.pdf")

# Show the plot
plt.show()
```
