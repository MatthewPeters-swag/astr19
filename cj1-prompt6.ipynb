# Defined sin

```python
import numpy as np

def my_sin(x):
    
    return np.sin(x)
```

# Defined cos

```python
def my_cos(x):
    
    return np.cos(x)
```

# Tabulate sin and cos
Tabulates between 2pi and 0 with a thousand entries

```python
x = np.linspace(0, 2, 1000)

sin_values = my_sin(x)
cos_values = my_cos(x)
```

# Print first 10 values
Prints first 10 values of sin and cos

```python
print("x         sin(x)    cos(x)")
print("----------------------------")

for i in range(10):
    print(f"{x[i]:.6f}  {sin_values[i]:.6f}  {cos_values[i]:.6f}")
```
