```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 1, 100)
```

```python
def sin_func(x):
    return np.sin(x)

def cos_func(x):
    return np.cos(x)
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))

# Left panel: sin(x)
ax[0].plot(x, sin_func(x))
ax[0].set_title("sin(x)")
ax[0].set_xlabel("x")
ax[0].set_ylabel("sin(x)")

# Right panel: cos(x)
ax[1].plot(x, cos_func(x))
ax[1].set_title("cos(x)")
ax[1].set_xlabel("x")
ax[1].set_ylabel("cos(x)")

plt.tight_layout()
plt.savefig("sin_cos.pdf")
plt.show()
```
